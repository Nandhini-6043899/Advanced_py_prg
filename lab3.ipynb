```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Inorder Traversal
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


# Preorder Traversal
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


# Postorder Traversal
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")


# Check Operator
def is_operator(value):
    return value in ['+', '-', '*', '/', '^']


# Construct Expression Tree from Postfix Expression
def construct_expression_tree(postfix):
    stack = []

    for token in postfix.split():
        if not is_operator(token):
            stack.append(Node(token))
        else:
            if len(stack) < 2:
                raise ValueError("Invalid postfix expression")

            operator_node = Node(token)
            operator_node.right = stack.pop()
            operator_node.left = stack.pop()
            stack.append(operator_node)

    if len(stack) != 1:
        raise ValueError("Invalid postfix expression")

    return stack.pop()


# Evaluate Expression Tree
def evaluate(root):
    if not is_operator(root.data):
        return float(root.data)

    left = evaluate(root.left)
    right = evaluate(root.right)

    if root.data == '+':
        return left + right
    elif root.data == '-':
        return left - right
    elif root.data == '*':
        return left * right
    elif root.data == '/':
        return left / right
    elif root.data == '^':
        return left ** right


# Main Program
print("==========================================")
print("      BINARY TREE AND EXPRESSION TREE")
print("==========================================")

# Create Binary Tree
root = Node(10)
root.left = Node(20)
root.right = Node(30)
root.left.left = Node(40)
root.left.right = Node(50)
root.right.right = Node(60)

# Binary Tree Traversals
print("\n--- BINARY TREE TRAVERSALS ---")

print("Inorder   : ", end="")
inorder(root)

print("\nPreorder  : ", end="")
preorder(root)

print("\nPostorder : ", end="")
postorder(root)

# Expression Tree
print("\n\n--- EXPRESSION TREE ---")

postfix = input("Enter postfix expression (Example: 3 5 + 2 *): ")

try:
    expression_root = construct_expression_tree(postfix)

    print("\nPostfix Expression:", postfix)
    print("Expression Tree Evaluation:", evaluate(expression_root))

except (IndexError, ValueError, ZeroDivisionError, OverflowError):
    print("Invalid expression!")
```

```python
class Queue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.queue = []

    def enqueue(self, item):
        if len(self.queue) == self.capacity:
            print("Queue Overflow!")
        else:
            self.queue.append(item)
            print(item, "inserted successfully.")

    def dequeue(self):
        if not self.queue:
            print("Queue Underflow!")
        else:
            item = self.queue.pop(0)
            print(item, "deleted successfully.")

    def peek(self):
        if not self.queue:
            print("Queue is empty.")
        else:
            print("Front Element:", self.queue[0])

    def display(self):
        if not self.queue:
            print("Queue is empty.")
        else:
            print("Queue:", self.queue)


class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.queue = [None] * capacity
        self.front = -1
        self.rear = -1

    def is_empty(self):
        return self.front == -1

    def is_full(self):
        return (self.rear + 1) % self.capacity == self.front

    def enqueue(self, item):
        if self.is_full():
            print("Circular Queue Overflow!")
            return

        if self.is_empty():
            self.front = 0

        self.rear = (self.rear + 1) % self.capacity
        self.queue[self.rear] = item
        print(item, "inserted successfully.")

    def dequeue(self):
        if self.is_empty():
            print("Circular Queue Underflow!")
            return

        item = self.queue[self.front]
        self.queue[self.front] = None

        if self.front == self.rear:
            self.front = -1
            self.rear = -1
        else:
            self.front = (self.front + 1) % self.capacity

        print(item, "deleted successfully.")

    def peek(self):
        if self.is_empty():
            print("Circular Queue is empty.")
        else:
            print("Front Element:", self.queue[self.front])

    def display(self):
        if self.is_empty():
            print("Circular Queue is empty.")
            return

        print("Circular Queue:", end=" ")
        i = self.front

        while True:
            print(self.queue[i], end=" ")
            if i == self.rear:
                break
            i = (i + 1) % self.capacity

        print()


# Main Program
print("======================================")
print("      QUEUE DATA STRUCTURE PROGRAM")
print("======================================")
print("1. Linear Queue")
print("2. Circular Queue")

queue_type = int(input("Select Queue Type: "))
capacity = int(input("Enter Queue Capacity: "))

if capacity <= 0:
    print("Queue capacity must be positive.")
    exit()

if queue_type == 1:
    q = Queue(capacity)
    queue_name = "LINEAR QUEUE"
elif queue_type == 2:
    q = CircularQueue(capacity)
    queue_name = "CIRCULAR QUEUE"
else:
    print("Invalid Queue Type!")
    exit()

while True:
    print("\n================================")
    print(" ", queue_name)
    print("================================")
    print("1. Enqueue")
    print("2. Dequeue")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            item = input("Enter element: ")
            q.enqueue(item)

        elif choice == 2:
            q.dequeue()

        elif choice == 3:
            q.peek()

        elif choice == 4:
            q.display()

        elif choice == 5:
            print("Program terminated.")
            break

        else:
            print("Invalid choice! Please try again.")

    except ValueError:
        print("Please enter a valid number.")
```

```python
class Stack:
    def __init__(self, capacity):
        self.capacity = capacity
        self.stack = []

    def is_empty(self):
        return len(self.stack) == 0

    def is_full(self):
        return len(self.stack) == self.capacity

    def push(self, item):
        if self.is_full():
            raise OverflowError("Stack Overflow: Stack is full.")
        self.stack.append(item)
        print(f"{item} pushed successfully.")

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack Underflow: Stack is empty.")
        item = self.stack.pop()
        print(f"{item} popped successfully.")
        return item

    def peek(self):
        if self.is_empty():
            raise IndexError("Stack is empty.")
        return self.stack[-1]

    def size(self):
        return len(self.stack)

    def display(self):
        if self.is_empty():
            print("Stack is empty.")
            return

        print("\nStack (TOP to BOTTOM):")
        for item in reversed(self.stack):
            print(item)


# Main Program
capacity = int(input("Enter stack capacity: "))
stack = Stack(capacity)

while True:
    print("\n===== STACK MENU =====")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Display")
    print("5. Size")
    print("6. Check Empty")
    print("7. Check Full")
    print("8. Exit")

    choice = input("Enter your choice: ")

    try:
        if choice == "1":
            item = input("Enter element: ")
            stack.push(item)

        elif choice == "2":
            stack.pop()

        elif choice == "3":
            print("Top Element:", stack.peek())

        elif choice == "4":
            stack.display()

        elif choice == "5":
            print("Stack Size:", stack.size())

        elif choice == "6":
            print("Stack Empty:", stack.is_empty())

        elif choice == "7":
            print("Stack Full:", stack.is_full())

        elif choice == "8":
            print("Program terminated.")
            break

        else:
            print("Invalid choice.")

    except (OverflowError, IndexError) as error:
        print(error)
```
